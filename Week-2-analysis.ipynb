Task 7: Create Employee Age
Think: How do you calculate age from date of birth?

```python
df[['DOB','StartDate']].head()
```

What this does:

Keeps only rows where DOB is present

Creates a fresh copy

Removes the warning permanently

```python
df=df[df['DOB'].notna()].copy()
```

```python
df['Age']=(pd.Timestamp('today')-df['DOB']).dt.days // 365

df[['DOB','Age']].head()
```

TASK 8: CREATE EMPLOYEE TENURE

Tenure = How long an employee has worked in the company

```python
df=df[df['StartDate'].notna()].copy()
```

```python
df['Tenure']=(pd.Timestamp('today')- df['StartDate']).dt.days // 365

df[['StartDate','Tenure']].head()
```

Task 9: Create grouped columns:

•	Age groups (example: 20–30, 30–40, etc.)

•	Tenure groups (example: 0–2 yrs, 2–5 yrs, etc.)


```python
df['Age-group']=pd.cut(
    df['Age'],
    bins=[18,25,35,45,55,65],
    labels=['18-25','26-35','36-45','46-55','56-65'],
    right=True             
)

df['Tenure-group']=pd.cut(
    df['Tenure'],
    bins=[0,1,5,10,20,50],
    labels=['0-1 yrs','2-5 yrs','6-10 yrs','11-20 yrs','20+ yrs'],
    right=True
)

df[['Age','Tenure','Age-group','Tenure-group']].head()
```

```python
df=df[df['Age-group'].notna()].copy()
```

TASK 20: FINAL INSIGHTS & BUSINESS RECOMMENDATIONS



1️⃣ Early-Tenure Attrition Is High

A significant portion of employee attrition occurs within the 0–1 year tenure group, indicating challenges in onboarding, role clarity, or early employee engagement.

2️⃣ Certain Departments Experience Higher Attrition

Departments such as Operations and Sales-oriented units show higher attrition compared to others, suggesting workload pressure or performance-linked stress.

3️⃣ Workforce Is Concentrated in Mid-Level Pay Zones

Most employees fall into mid-level pay zones, indicating a cost-balanced workforce structure but limited representation at senior compensation levels.

4️⃣ Internal Training Programs Show Better Completion Rates

Internal training programs have a higher completion rate than external programs, making them more reliable for skill development.

5️⃣ High Training Cost Does Not Always Guarantee Better Outcomes

Some training programs with higher average costs do not show proportionally better completion or success rates, pointing to inefficiencies in training investments.

6️⃣ Training Effectiveness Varies by Employee Experience

Employees with 2–5 years of tenure tend to complete training programs more successfully compared to new joiners, highlighting the importance of experience in learning outcomes.

7️⃣ Gender and Employment Type Distribution Is Uneven

The workforce shows an imbalance across gender and employment types, indicating scope for improving diversity and inclusion initiatives.



💡 BUSINESS RECOMMENDATIONS (ACTIONABLE)

✅ 1️⃣ Strengthen Onboarding & First-Year Engagement

Introduce structured onboarding programs, mentorship, and regular check-ins during the first year to reduce early-stage attrition.

✅ 2️⃣ Department-Specific Retention Strategies

Conduct focused HR interventions in high-attrition departments, such as workload balancing, role clarity, and manager training.

✅ 3️⃣ Optimize Training Budget Allocation

Re-evaluate high-cost, low-impact training programs and redirect investment toward programs with proven completion and success rates.

✅ 4️⃣ Expand and Improve Internal Training Programs

Since internal training shows higher effectiveness, increase internal knowledge-sharing sessions, in-house workshops, and peer learning initiatives.

✅ 5️⃣ Customize Training Based on Tenure Level

Design separate training paths for new joiners and experienced employees to improve engagement and learning outcomes.

✅ 6️⃣ Improve Workforce Diversity & Inclusion

Implement targeted hiring and HR policies to improve gender balance and employment type diversity, creating a more inclusive workplace.

✅ 7️⃣ Use Data-Driven HR Monitoring

Establish periodic HR dashboards to continuously track attrition, training performance, and cost efficiency, enabling proactive decision-making.
